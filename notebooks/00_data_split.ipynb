```python
df['date'] = pd.to_datetime(df['date'])
print(df['date'].dt.year.value_counts())
```

**Avoid Leakage**
1. Your train set (2012-2019) is "clean" historical training.
2. Your eval set (2020-2021) is "unseen" during training, but used for validation and hyperparameter tuning.
3. Your holdout set (2022-2023) is completely untouched until the very end. 

```python
# sort by date (important for time series data)
df = df.sort_values('date')
```

```python
# define cutoff dates
cutoff_date_eval = '2020-01-01' # validation starts
cutoff_date_holdout = '2022-01-01' # holdout starts
```

```python
# Train: before 2020-01-01
df_train = df[df['date'] < cutoff_date_eval]

# Validation: between 2020-01-01 and 2022-01-01
df_eval = df[(df['date'] >= cutoff_date_eval) & (df['date'] < cutoff_date_holdout)]

# Holdout: after 2022-01-01
df_holdout = df[df['date'] >= cutoff_date_holdout]

print(f"Train set: {df_train.shape}")
print(f"Eval set: {df_eval.shape}") 
print(f"Holdout set: {df_holdout.shape}")   
```

```python
# save the splits
df_train.to_csv(r'E:\End-to-end-machine-learning-project-work-flow\data\raw\train.csv', index=False)
df_eval.to_csv(r'E:\End-to-end-machine-learning-project-work-flow\data\raw\eval.csv', index=False)
df_holdout.to_csv(r'E:\End-to-end-machine-learning-project-work-flow\data\raw\holdout.csv', index=False)
```
