```python
#======================================
# 01. Imports
#======================================
import pandas as pd
import numpy as np
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error , mean_squared_error, r2_score
```

```python
#======================================
# 02. Loading datasets
#======================================
# Use train.csv for training (2012-2019 or 2012-2020 depending on eval split)
# Use eval.csv (2020-2021) for validation
# Keep holdout.csv (2022) untouched for final testing

train_df = pd.read_csv(r'E:\End-to-end-machine-learning-project-work-flow\data\processed\train_fe.csv')
eval_df = pd.read_csv(r'E:\End-to-end-machine-learning-project-work-flow\data\processed\eval_fe.csv')
```

```python
# =======================================
# 03. Define target & features
# =======================================

target = 'price'
X_train = train_df.drop(columns=[target])
y_train = train_df[target]

X_eval = eval_df.drop(columns=[target])
y_eval = eval_df[target]
```

```python
#======================================
# 04. Baseline model (Dummy Regressor)
#======================================
baseline = DummyRegressor(strategy='median')
baseline.fit(X_train, y_train)
```

```python
#======================================
# 05. Predictions & Metrics
#======================================
y_pred = baseline.predict(X_eval)

mae = mean_absolute_error(y_eval, y_pred)
rmse = np.sqrt(mean_squared_error(y_eval, y_pred))
r2 = r2_score(y_eval, y_pred)

print("Baseline Dummy Regressor (Median) Performance:")
print(f"Mean Absolute Error: {mae:,.2f}")
print(f"Root Mean Squared Error: {rmse:,.2f}")
print(f"R^2 Score: {r2:.4f}")
```
